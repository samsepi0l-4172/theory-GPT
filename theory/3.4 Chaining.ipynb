```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain.callbacks import StreamingStdOutCallbackHandler


class ChefChain:
    """
    Represents a chain of prompts and chat responses for a world-class international chef.

    Attributes:
        chat (ChatOpenAI): An instance of the ChatOpenAI class for generating chat responses.
        chef_prompt (ChatPromptTemplate): A template for the initial prompt in the chat.
    """

    def __init__(self):
        self.chat = ChatOpenAI(
            temperature=0.1,
            streaming=True,
            callbacks=[StreamingStdOutCallbackHandler()],
        )

        self.chef_prompt = ChatPromptTemplate.from_messages(
            [
                (
                    "system",
                    "You are a world-class international chef. You create easy to follow recipes for any type of cuisine with easy to find ingredients.",
                ),
                ("human", "I want to cook {cuisine} food."),
            ]
        )

    def create_chef_chain(self):
        """
        Creates a chain of prompts and chat responses for a world-class international chef.

        Returns:
            str: The generated chat response.
        """
        chef_chain = self.chef_prompt | self.chat
        return chef_chain


# Create an instance of the ChefChain class
chef_chain_instance = ChefChain()

# Generate the chef chain
chef_chain_response = chef_chain_instance.create_chef_chain()
```

```python
def invoke_veg_chain(recipe: str, cuisine: str) -> str:
    """
    Invoke the final chain of the vegetarian chef prompt and the chat prompt.

    Args:
        recipe (str): The recipe to be passed to the vegetarian chef prompt.
        cuisine (str): The cuisine to be passed to the final chain.

    Returns:
        str: The output generated by the final chain.
    """
    veg_chef_prompt = ChatPromptTemplate.from_messages(
        [
            (
                "system",
                "You are a vegetarian chef specialized in making traditional vegetarian recipes. You find alternative ingredients and explain their preparation. You don't radically modify the recipe. If there is no alternative for a food, just say you don't know how to replace it.",
            ),
            ("human", recipe),
        ]
    )

    veg_chain = veg_chef_prompt | chat

    final_chain = {"recipe": chef_chain} | veg_chain

    return final_chain.invoke({"cuisine": cuisine})


veg_chef_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a vegetarian chef specialized on making traditional vegeterian recipies. You find alternative ingredients and explain their preparation. You don't radically modify the recipe. If there is no alternative for a food just say you don't know how to replace it.",
        ),
        ("human", "{recipe}"),
    ]
)

veg_chain = veg_chef_prompt | chat

final_chain = {"recipe": chef_chain} | veg_chain

final_chain.invoke({"cuisine": "indian"})
```
