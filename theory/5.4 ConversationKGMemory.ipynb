```python
from langchain_community.memory.kg import ConversationKGMemory
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(temperature=0.1)

memory = ConversationKGMemory(
    llm=llm,
    return_messages=True,
)


def add_message(input, output):
    """
    Saves the input and output messages to the memory context.

    Parameters:
    input (str): The input message to be saved.
    output (str): The output message to be saved.

    Returns:
    None
    """
    memory.save_context({"input": input}, {"output": output})


add_message("Hi I'm samspei0l, I live in South Korea", "Wow that is so cool!")
```

```python
memory.load_memory_variables({"input": "Who is samspei0l?"})
```

```python
add_message("samsepi0l likes kimchi", "Wow that is so cool!")
```

```python
memory.load_memory_variables({"inputs": "what does samsepi0l like"})
```
