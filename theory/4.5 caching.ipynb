```python
from langchain_community.cache import InMemoryCache, SQLiteCache
from langchain_core.callbacks import StreamingStdOutCallbackHandler
from langchain_core.globals import set_debug, set_llm_cache
from langchain_openai import ChatOpenAI


def run_chat():
    """
    Run the chat application using the ChatOpenAI model.

    This function initializes the necessary components for the chat application,
    sets the cache and debug options, and predicts a response to the input query.

    Args:
        None

    Returns:
        None
    """
    # set_llm_cache(InMemoryCache())
    set_llm_cache(SQLiteCache("cache.db"))
    set_debug(True)

    chat = ChatOpenAI(
        temperature=0.1,
        # streaming=True,
        # callbacks=[
        #     StreamingStdOutCallbackHandler(),
        # ],
    )

    chat.predict("How do you make Italian pasta")


run_chat()
```
