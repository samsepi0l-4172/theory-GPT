```python
from langchain_core.callbacks import StreamingStdOutCallbackHandler
from langchain_core.prompts import PipelinePromptTemplate, load_prompt
from langchain_openai import ChatOpenAI


def process_chat():
    """
    This function initializes a chatbot using the ChatOpenAI class and processes the chat using the provided prompt.

    Returns:
        None
    """
    prompt = load_prompt("./prompt.yaml")

    chat = ChatOpenAI(
        temperature=0.1,
        streaming=True,
        callbacks=[
            StreamingStdOutCallbackHandler(),
        ],
    )

    prompt.format(country="Germany")


process_chat()
```
