```python
from typing import List
from pydantic.v1 import BaseModel, Field


class radio_answer_schema(BaseModel):
    answer: str = Field(description="function that make a radio answer")
    correct: bool = Field(description="function that is the answer correct")


class radio_schema(BaseModel):
    """Make a radio questions and answers and returns a quiz"""

    question: str = Field(
        description="function that takes a string of question and return a quiz"
    )
    answers: List[radio_answer_schema] = Field(
        description="function that takes a list and answers and returns a quiz"
    )
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI()
```

```python
from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser(pydantic_object=radio_schema)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template="Make a 10 quizes about {city}",
    input_variables=["query"],
    partial_variables={
        "format_instructions": parser.get_format_instructions(),
    },
)

chain = prompt | llm | parser

response = chain.invoke(
    {
        "city": "rome",
    }
)

# response = response.additional_kwargs["function_call"]["arguments"]

print(response)
```
